QUES:
What is the difference between a function and a method in Python?

SOL:
A function is a block of reusable code that is defined using the def keyword. It can be called independently.

A method is like a function, but it is associated with an object. It is called using the object, and often acts on that object.

```python
# Function
def greet(name):
    return "Hello, " + name

print(greet("Anita"))

# Method
name = "Anita"
print(name.upper())  # upper() is a string method
```

QUES:
Explain the concept of function arguments and parameters in Python.

SOL:
Parameters are variables listed in the function definition.

Arguments are the values passed to the function when calling it.

```python
def add(a, b):  # a and b are parameters
    return a + b

result = add(5, 3)  # 5 and 3 are arguments
print(result)
```

You can define functions in these ways:

QUES: What are iterators in Python and how do they differ from iterables?

SOL:
Iterable: An object that can return an iterator (like lists, strings).

Iterator: An object with a __next__() method, which returns elements one by one.

```python
# Iterable
numbers = [1, 2, 3]
it = iter(numbers)  # Creating an iterator from the list

print(next(it))
print(next(it))
print(next(it))
```

QUES:
Explain the concept of generators in Python and how they are defined.

SOL:
Generators are functions that yield values one by one using the yield keyword. They don’t store all values in memory.

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

for i in countdown(3):
    print(i)
```

QUES:
What are the advantages of using generators over regular functions?

SOL:
1. Save memory because they don't store all items at once.

2. They generate items only when needed (lazy evaluation).

3. Useful for large data processing.

```python
def big_range():
    for i in range(1000000):
        yield i

gen = big_range()
print(next(gen))
print(next(gen))
```

QUES: Explain the purpose and usage of the map() function in Python.

SOL:
map() applies a function to every item of an iterable (like a list). It returns a map object which can be converted to a list.

```python
numbers = [1, 2, 3, 4]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
```

QUES:
What is the difference between map(), reduce(), and filter() functions in Python?

SOL:
1. map() applies a function to all elements.

2. filter() filters elements based on a condition.

3. reduce() combines all elements into a single value (requires importing from functools).

```python
from functools import reduce

numbers = [1, 2, 3, 4]

# map
squared = list(map(lambda x: x ** 2, numbers))
print("map:", squared)

# filter
even = list(filter(lambda x: x % 2 == 0, numbers))
print("filter:", even)

# reduce
sum_all = reduce(lambda x, y: x + y, numbers)
print("reduce:", sum_all)
```

QUES:
Using pen & paper write the internal mechanism for sum operation using reduce function on this list: [47,11,42,13]

SOL:
Using reduce(lambda x, y: x + y, [47,11,42,13]), it works like this:
Step 1: 47 + 11 = 58
Step 2: 58 + 42 = 100
Step 3: 100 + 13 = 113


```python
from functools import reduce

nums = [47, 11, 42, 13]
result = reduce(lambda x, y: x + y, nums)
print("Sum using reduce:", result)
```
