### 1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

```python
def sum_of_evens(numbers):
    numb = []
    for num in numbers:
        if num % 2 == 0:
            numb.append(num)
    return sum(numb)
```

```python
sum_of_evens([1,2,3,4])
```

### 2. Create a Python function that accepts a string and returns the reverse of that string.

```python
def rev_str(a):
    if type(a) == str:
        return a[::-1]
```

```python
rev_str("shivam")
```

### 3. Implement a Python function that takes a list of integers and returns a new list containing the squares of each number.

```python
def int_sqr(a):
    n = []
    for i in a:
        if type(i) == int:
            n.append(i*i)
    return n
```

```python
int_sqr([1,2,3,"Milk"])
```

### 4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
for num in range(1, 201):
    print(f"{num} is {'Prime' if is_prime(num) else 'Not Prime'}")
```

### 5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

```python
def fibo(n):
    a = 0
    b = 1
    for i in range(n):
        yield a
        a, b = b, a+b
```

```python
for i in fibo(10):
    print(i)
```

### 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def powers_of_two(expo):
    for i in range(expo + 1):
        yield 2 ** i
```

```python
for p in powers_of_two(10):
    print(p)
```

### 7. Implement a generator function that reads a file line by line and yields each line as a string.

```python
file = ["I am a student.", "I am learning python.", "I am studying to become data analyst."]
```

```python
def file_reader(file):
    for line in file:
        yield str(line.strip())
```

```python
list(file_reader(file))
```

### 8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

```python
# This is the sample data.
data = [(1, 5), (3, 2), (4, 9), (2, 1)]
```

```python
# Sorting using the lambda function
sorted_data = sorted(data, key=lambda x: x[1])
```

```python
# Output based on second element of the tuples.
print(sorted_data)
```
